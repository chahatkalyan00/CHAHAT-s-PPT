```python
text = input("Enter text: ").lower()
words = text.split()

word_list = []
word_count = []

for w in words:
    if w in word_list:
        idx = word_list.index(w)
        word_count[idx] = word_count[idx] + 1
    else:
        word_list.append(w)
        word_count.append(1)

print("Word Frequency:")
for i in range(len(word_list)):
    print(word_list[i], "=", word_count[i])
```
